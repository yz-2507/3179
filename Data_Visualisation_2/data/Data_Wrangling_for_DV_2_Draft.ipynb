```python
Full_Obesity = pd.read_csv('Full_obesity_data.csv')
```

```python
Full_Obesity
```

```python
australia_obesity = Full_Obesity.loc[(Full_Obesity['Country']=='Australia'),['Country','Year','Obesity (%)','Sex']]
```

```python
australia_obesity = australia_obesity.dropna()
```

```python
australia_obesity
```

```python
def retrieve_val(x):
    val_lst = x.split('[')
    return val_lst[0]
```

```python
australia_obesity['Obesity (%)'] = australia_obesity['Obesity (%)'].apply(retrieve_val)
```

```python
australia_obesity.to_csv('Australia_obesity_data.csv')
```
