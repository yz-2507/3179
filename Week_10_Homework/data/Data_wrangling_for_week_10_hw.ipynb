```python
import pandas as pd
```

```python
BMI_male_df = pd.read_csv('BMI_Male.csv')
```

```python
Obesity_df = pd.read_csv('Obesity_data.csv')
```

```python
Population_df = pd.read_csv('Population_data.csv')
```

```python
Continent_df = pd.read_csv('Continents_data.csv')
```

```python
filtered_BMI_male_df = BMI_male_df.loc[(BMI_male_df['Year'] == 2016), ['Entity','Mean BMI (male)']]
```

```python
filtered_BMI_male_df.rename(columns={'Entity':'Country'},inplace=True)
```

```python
filtered_BMI_male_df
```

```python
Obesity_df
```

```python
Population_df = Population_df.dropna()
```

```python
Population_df['2016'] = Population_df['2016'].astype(int)
```

```python
Population_df.rename(columns={'Country Name':'Country','2016':'Population_2016'},inplace=True)
```

```python
Population_df
```

```python
merge_df = pd.merge(filtered_BMI_male_df,Obesity_df, on='Country',how='inner')
```

```python
merge_df = pd.merge(merge_df,Population_df, on = 'Country', how = 'inner')
```

```python
Continent_df
```

```python
merge_df = pd.merge(merge_df,Continent_df, on = 'Country', how = 'inner')
```

```python
merge_df
```

```python
merge_df.to_csv('Male_BMI_Obesity.csv')
```
